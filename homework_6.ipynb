## 1. Непрерывная возрастающая последовательность
Если следующий элемент больше предыдущего, увеличиваем `cur`, иначе начинаем заново. В `best` храним максимум. **Время O(n), память O(1)**.

```python
def longest_increasing_run(nums):
    if not nums:
        return 0

    cur = best = 1
    for i in range(1, len(nums)):
        if nums[i] > nums[i - 1]:
            cur += 1
        else:
            cur = 1
        best = max(best, cur)
    return best


assert longest_increasing_run([3, 2, 8, 9, 5, 10]) == 3
assert longest_increasing_run([1, 2, 7, 9, 0, 10]) == 4
assert longest_increasing_run([8, 8, 8, 8]) == 1
assert longest_increasing_run([]) == 0
print(longest_increasing_run([3, 2, 8, 9, 5, 10]))
```

## 2. Треугольник Паскаля
По краям каждой строки ставим 1, остальные числа — сумма двух соседних из предыдущей строки. **Время O(n²), память O(n²)** для результата.

```python
def pascal_triangle(n):
    triangle = []
    for i in range(n):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[-1][j - 1] + triangle[-1][j]
        triangle.append(row)
    return triangle


assert pascal_triangle(0) == []
assert pascal_triangle(1) == [[1]]
assert pascal_triangle(5) == [[1], [1, 1], [1, 2, 1], [1, 3, 3, 1], [1, 4, 6, 4, 1]]
print(pascal_triangle(5))
```

## 3. Размен монет
`dp[s]` — минимальное количество монет для суммы `s`. Пробуем каждый номинал: `dp[s] = min(dp[s], dp[s - coin] + 1)`. Если сумма недостижима, возвращаем -1. **Время O(amount · len(coins)), память O(amount)**.

```python
def coin_change(coins, amount):
    if amount < 0:
        return -1

    dp = [amount + 1] * (amount + 1)
    dp[0] = 0

    for s in range(1, amount + 1):
        for coin in coins:
            if 0 < coin <= s:
                dp[s] = min(dp[s], dp[s - coin] + 1)

    return dp[amount] if dp[amount] != amount + 1 else -1


assert coin_change([1, 2, 5], 11) == 3
assert coin_change([2], 3) == -1
assert coin_change([1], 0) == 0
assert coin_change([], 7) == -1
assert coin_change([2, 5], 6) == 3
print(coin_change([1, 2, 5], 11))
```

## 4. Самый длинный палиндром
Расширяемся от каждого центра, пока символы слева и справа равны. Проверяем центры из одного и двух символов. Запоминаем самую длинную подстроку. **Время O(n²), память O(1)** без учёта ответа.

```python
def longest_palindrome(s):
    left_best = right_best = 0

    def expand(left, right):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return left + 1, right

    for i in range(len(s)):
        for left, right in (expand(i, i), expand(i, i + 1)):
            if right - left > right_best - left_best:
                left_best, right_best = left, right

    return s[left_best:right_best]


assert longest_palindrome('babad') in ('bab', 'aba')
assert longest_palindrome('cbbd') == 'bb'
assert longest_palindrome('racecar') == 'racecar'
assert longest_palindrome('abc') in ('a', 'b', 'c')
assert longest_palindrome('') == ''
print(longest_palindrome('babad'))
```
